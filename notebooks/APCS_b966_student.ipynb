### 題目規則（教學改寫）

- 第一個整數 `N` 是線段數量；接下來 N 列，每列有左端點 `L`、右端點 `R`，且 `L <= R`。
- **同一位置被多條線段蓋住，只計一次長度。** 例如 `(1, 5)` 和 `(3, 7)` 的總覆蓋長度是 `6`，不是 `4 + 4 = 8`。
- 範例覆蓋示意：`(1,2)、(4,8)、(5,6)、(7,9)` 合起來是 `(1,2)` 與 `(4,9)`，長度 `1 + 5 = 6`。
- 測資最高等級：`N < 10000`，座標在 `0` 到 `10000000` 以下。需要有效率的排序與掃描方式。

**輸入**
```text
5

150 200
280 300
300 330
190 210
```
**輸出**
```text
110
```

因為合併後是 `(150,210)` 與 `(280,330)`，所以 `60 + 50 = 110`。

### 官方範例二

**輸入**
```text

```
**輸出**
```text

```

**注意**：連在一起的 `(2,4)` 與 `(4,6)` 可以合併為 `(2,6)`；端點本身不會額外增加長度。

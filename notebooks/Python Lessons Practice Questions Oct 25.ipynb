一行若干個以空格分隔的整數。

````python
def count_even(numbers):
    count=0
    for i in range(len(numbers)):
        if numbers[i]%2==0:
            count+=1
    return count

numbers= input(). split()
numbers= list(int(x) for x in numbers)
ans= count_even(numbers)
print(ans)
````

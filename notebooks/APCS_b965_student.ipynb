### 兩種正向操作（A → B）

| 代碼 | 正向動作 | 還原時的逆動作 |
|---|---|---|
| `0` | **順時針旋轉 90°** | **逆時針旋轉 90°** |
| `1` | **上下翻轉**（第一列與最後一列交換） | **上下翻轉** |

**關鍵陷阱**：還原 B 時需要**倒過來處理指令**。

輸入：`R C M`，接著 R 列每列 C 個數字，接著 M 個操作（`0` 或 `1`）。輸出 A 的列數、行數，再列出 A 的每一列；本題為嚴格格式比對。

````python
# === 練習題自動檢查工具：先執行，請不要修改 ===
def _check(title, func, cases):
    count = 0
    for i, (args, expected) in enumerate(cases, 1):
        try:
            got = func(*args)
            ok = got == expected
        except Exception as error:
            ok, got = False, f"發生錯誤：{type(error).__name__}: {error}"
        print(f"{'✅ PASS' if ok else '❌ 再試試'} | {title} 測試 {i}")
        if not ok:
            print(f"   輸入參數：{args!r}")
            print(f"   預期：{expected!r}")
            print(f"   實際：{got!r}")
        count += ok
    print(f"{title} 完成度：{count}/{len(cases)}\n")
    return count
````

## 1. 基本 `for` 迴圈

如果希望同一段程式重複執行多次，可以搭配：

```python
for i in range(...):
    ...
```

下面的程式會重複印出 `"hi"`。


````python
for i in range(30):
    print("hi")
````

## 2. `range(5)` 與迴圈變數

`range(5)` 會依序產生：

```text
0, 1, 2, 3, 4
```

下面的例子也示範：即使在迴圈內把 `i` 改成 `20`，下一輪 `for` 還是會從 `range(5)` 取得下一個值。


````python
for i in range(5):  # i = 0 -> i = 1 -> ...
    print(i)
    i = 20
````

### 把 `range()` 轉成 List

可以使用：

```python
list(range(5))
```

直接觀察 `range(5)` 產生的內容。


````python
list(range(5))  # -> 可迭代 (iterable)
````

## 3. `for` 可以逐一讀取 List

`for` 不只能搭配 `range()`。

List 也是 iterable，因此可以直接寫：

```python
for i in lst:
    ...
```

每一輪會依序取得 List 中的一個元素。


````python
lst = [0, 1, 2, 3, 4, "a", "c", [23, 78]]

for i in lst:
    print(i)
````

## 4. `range(start, stop, step)`

`range()` 可以指定：

```python
range(頭, 尾, 步長)
```

其中 `stop` **不包含在結果中**。

例如：

```text
range(1, 5)    -> 1, 2, 3, 4
range(1, 5, 2) -> 1, 3
range(2, 19, 4) -> 2, 6, 10, 14, 18
range(10)      -> range(0, 10)
```


````python
# 0 - 10 => (1, 3, 5, 7, 9)
for i in range(1, 10, 2):
    print(i)
````

````python
# range(頭, 尾, 步長) -> 留頭不留尾，步長就是每次前進的距離
# range(1, 5) -> 1, 2, 3, 4
# range(1, 5, 2) -> 1, 3
# range(2, 19, 4) -> 2, 6, 10, 14, 18
# range(10) -> range(0, 10)
````

## 5. `input().split()`

如果輸入：

```text
1 2 3 4 5
```

執行：

```python
input().split()
```

會先得到由**字串**組成的 List。


````python
a = input().split()  # 例如輸入：1 2 3 4 5

print(a)
````

### 逐一轉成整數

原始 notebook 使用 `count` 記錄目前的 index，再把每一個元素轉成 `int`。

流程可以理解成：

```text
a[0] -> int(a[0])
a[1] -> int(a[1])
a[2] -> int(a[2])
...
```


````python
count = 0

for i in a:
    a[count] = int(a[count])
    count += 1

print(a)
````

### 使用 `len()`

`len(a)` 可以取得 List 中目前有多少個元素。


````python
len(a)
````

## 6. 使用 `range(len(list))` 取得 Index

如果除了元素本身之外，也需要知道它的位置，可以使用：

```python
for i in range(len(list1)):
    ...
```

此時 `i` 會是：

```text
0, 1, 2, ...
```

因此可以用 `list1[i]` 或其他 List 的 `[i]` 存取相同位置的資料。


````python
# 轉成字串，讓後面的 lst_str 第 i 個對應 list1 的第 i 個

list1 = [1, 2, 3, 4, 5]
lst_str = ["first", "second", "third", "fourth", "fifth"]

for i in range(len(list1)):
    print(f"list1 的第 {i + 1} 個是 {lst_str[i]}")
````

## 7. 練習：依座號分組

電腦課要同學分組做期末報告。

分組方式為依座號順序，每 3 個人一組：

```text
1, 2, 3   -> 第 1 組
4, 5, 6   -> 第 2 組
7, 8, 9   -> 第 3 組
...
```

輸入同學的座號，判斷他在哪一組。

原始程式先把座號減 1，再使用整數除法：

```python
// 3
```

計算組別。


````python
x = input()
x = int(x) - 1

# 1, 2, 3, 4, ...
# ↓ 全部先減 1
# 0, 1, 2 -> // 3 = 0 -> 0 + 1

print(x // 3 + 1)
````

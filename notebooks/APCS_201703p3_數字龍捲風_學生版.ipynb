## 原題整理

給定邊長為奇數的 `N×N` 矩陣（每格 0～9），從**正中心**出發，向指定方向走第一步，接著**每完成一段順時針轉向**。每兩段才把步長加 1：`1,1,2,2,3,3,…`。走訪每格恰好一次，將數字**直接串接、不加空白**輸出。

| 方向編號 | 方向 | `(dr,dc)` |
|---|---|---|
| `0` | 左 | `(0,-1)` |
| `1` | 上 | `(-1,0)` |
| `2` | 右 | `(0,1)` |
| `3` | 下 | `(1,0)` |

| 子題組 | 分數 | 限制 |
|---|---:|---|
| 1 | 20 分 | `3≤N≤5`，起始方向 `0`（左） |
| 2 | 80 分 | `3≤N≤49`，四方向皆可能 |

**題目範例一**

```text
5
0
3 4 2 1 4
4 2 3 8 9
2 1 9 5 6
4 2 3 7 8
1 2 6 4 3
```

預期輸出：`9123857324243421496834621`

**題目範例二**

```text
3
1
4 1 2
3 0 5
6 7 8
```

預期輸出：`012587634`

## 學習方式

先執行共用檢查格，接著依序完成每題的 TODO。**每題下方立即附有可重複執行的獨立 PASS/FAIL 自動檢查格**。尚未完成的題目會顯示 `FAIL`，不會中斷後面的練習。

注意：兩份教材中的測試結果都是**本機檢查**，不是線上評測的 AC 證明。


````python
import copy
TEST_RESULTS = {}

def check(name, fn, cases):
    """每題獨立檢查：TODO / 執行錯誤 / 改動輸入都回報 FAIL。"""
    passed = 0
    failures = []
    for number, (args, expected) in enumerate(cases, 1):
        inputs = copy.deepcopy(args)
        before = copy.deepcopy(inputs)
        try:
            actual = fn(*inputs)
            correct = (type(actual) is type(expected) and
                       actual == expected and inputs == before)
        except NotImplementedError as error:
            actual = '尚未完成 TODO：'+str(error)
            correct = False
        except Exception as error:
            actual = type(error).__name__ + ': ' + str(error)
            correct = False
        if correct:
            passed += 1
        elif len(failures) < 2:
            failures.append((number,repr(args)[:100],repr(expected)[:110],repr(actual)[:110]))
    TEST_RESULTS[name] = (passed, len(cases))
    print('PASS' if passed == len(cases) else 'FAIL',name,str(passed)+'/'+str(len(cases)))
    for number, args, expected, actual in failures:
        print('  案例',number,'輸入:',args,'預期:',expected,'實際:',actual)
    return passed == len(cases)

def group_report(group,names):
    num=0
    for name in names:
        if name in TEST_RESULTS:
            passed,total=TEST_RESULTS[name]
            if passed==total:
                num+=1
    print(group+' 完成：',num,'/',len(names),'題')

def final_report():
    names=['Q%02d'%i for i in range(1,13)]
    group_report('全課總結',names)
````

## Q01｜找出正中央格子

**任務：** 給定奇數 `n`，回傳矩陣中心的 `(列, 欄)`。例如 `n=5` → `(2,2)`。

**函式：** `q1_center(n)`

**提示：** 索引從 0 開始；兩個座標都可使用 `n//2`。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q1_center(n):
    # TODO Q01：找出正中央格子
    r = n // 2
    return (r,r)
````

````python
# Q01｜獨立 PASS/FAIL 自動檢查（24 組）
CASES_Q01 = [
    ((3,), (1, 1)),
    ((5,), (2, 2)),
    ((7,), (3, 3)),
    ((9,), (4, 4)),
    ((11,), (5, 5)),
    ((13,), (6, 6)),
    ((15,), (7, 7)),
    ((17,), (8, 8)),
    ((19,), (9, 9)),
    ((21,), (10, 10)),
    ((23,), (11, 11)),
    ((25,), (12, 12)),
    ((27,), (13, 13)),
    ((29,), (14, 14)),
    ((31,), (15, 15)),
    ((33,), (16, 16)),
    ((35,), (17, 17)),
    ((37,), (18, 18)),
    ((39,), (19, 19)),
    ((41,), (20, 20)),
    ((43,), (21, 21)),
    ((45,), (22, 22)),
    ((47,), (23, 23)),
    ((49,), (24, 24)),
]
check('Q01', q1_center, CASES_Q01)
````

## Q02｜將方向編號轉成位移

**任務：** 依照 `0=左、1=上、2=右、3=下`，回傳 `(dr,dc)`。

**函式：** `q2_delta(d)`

**提示：** 往上是列減少；往左是欄減少。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q2_delta(d):
    # TODO Q02：將方向編號轉成位移
    dr = [0, -1, 0, 1]
    dc = [-1, 0, 1, 0]
    return (dr[d],dc[d])
````

````python
# Q02｜獨立 PASS/FAIL 自動檢查（4 組）
CASES_Q02 = [
    ((0,), (0, -1)),
    ((1,), (-1, 0)),
    ((2,), (0, 1)),
    ((3,), (1, 0)),
]
check('Q02', q2_delta, CASES_Q02)
````

## Q03｜每完成一段就順時針右轉

**任務：** 給定目前方向 `d`，回傳順時針旋轉 90 度後的編號。

**函式：** `q3_turn(d)`

**提示：** `0→1→2→3→0`，可使用 `%4`。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q3_turn(d):
    # TODO Q03：每完成一段就順時針右轉
    # q3(0) -> 1 ... q3(3) -> 0
    return (d+1) % 4
````

````python
# Q03｜獨立 PASS/FAIL 自動檢查（4 組）
CASES_Q03 = [
    ((0,), 1),
    ((1,), 2),
    ((2,), 3),
    ((3,), 0),
]
check('Q03', q3_turn, CASES_Q03)
````

## Q04｜產生每段移動步數

**任務：** 給定段數 `k`，回傳每段應走步數的清單。例如 `k=7` → `[1,1,2,2,3,3,4]`。

**函式：** `q4_steps(k)`

**提示：** 每兩段步數才增加 1；`k=0` 時回傳空串列。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q4_steps(k):
    # TODO Q04：產生每段移動步數
    ans = []
    length = 1
    for segment in range(k):
        ans.append(length)
        if segment %2 == 1:
            length += 1
    return ans
````

## Q05｜根據方向向前移動一格

**任務：** 傳入目前 `(r,c)` 與方向 `d`，回傳向前一步的新座標。**本題只練位移，暫時不檢查邊界。**

**函式：** `q5_move(r,c,d)`

**提示：** 先取得 `(dr,dc)`，再加到舊座標。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q5_move(r,c,d):
    # TODO Q05：根據方向向前移動一格
    dr = [0, -1, 0, 1]
    dc = [-1, 0, 1, 0]
    return (r + dr[d], c + dc[d])
````

## Q06｜走完同一方向的一整段

**任務：** 從 `start=(r,c)` 出發，往 `d` 方向走 `length` 步，回傳**走過的每個新座標**（不含起點）。

**函式：** `q6_segment(start,d,length)`

**提示：** 先更新座標，再把每一步的位置 `append` 進結果。這題只計算座標，先不檢查矩陣邊界。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q6_segment(start,d,length):
    # TODO Q06：走完同一方向的一整段
    r,c = start
    
    dr = [0, -1, 0, 1]
    dc = [-1, 0, 1, 0]

    ans = []

    for i in range(length):
        r += dr[d]
        c += dc[d]
        ans.append((r,c))
    
    return ans
````

## Q07｜產生完整螺旋座標順序

**任務：** 從 `(n//2,n//2)` 開始，把 `n*n` 個座標依走訪順序放入串列。起點要放第一個；每段完成轉向；滿 `n*n` 個座標**立即停止**。

**函式：** `q7_positions(n,d)`

**提示：** 使用 `while len(path)<n*n`，外層控制步長，內層連走兩段。

**動手前想一想：** 先用紙筆寫出一組小測資的預期結果，再執行下方的自動檢查。

````python
def q7_positions(n,d):
    # TODO Q07：產生完整螺旋座標順序
    r, c = n//2, n//2
    ans = [(r,c)]

    dr = [0, -1, 0, 1]
    dc = [-1, 0, 1, 0]

    length = 1

    while len(ans) < n*n:
        for segment in range(2):

            for i in range(length):
                if len(ans) == n*n:
                    break
                r += dr[d]
                c += dc[d]
                ans.append((r,c))

            d = (d+1) % 4
        length += 1
    return(ans)
````

## 學生提交前檢查

完成 Q12 並通過檢查後，請把 **Q12 的核心模擬演算法**改寫成從 `input()` 讀取資料、使用 `print()` 輸出一行字串的獨立程式。不要貼入自動檢查格，也不能在評測輸出中出現 `PASS/FAIL` 或示範訊息。

## 課堂總結：常見錯誤

1. 把方向代碼當成 `上右下左`；題目明確是 `左上右下`。
2. 忘記把中心格列入答案，導致長度少一。
3. 把步長寫成 `1,2,3,4,…`，正確是 `1,1,2,2,3,3,…`。
4. 每走一格就轉向，而不是每完成一段才轉。
5. 最後一段走過頭越界；應在恰好走完 `N*N` 格時停止。
6. 使用 `print(x,end=" ")` 產生空白；本題要連接數字輸出。

**進階思考：** 時間複雜度 `O(N²)`，因為每個矩陣格子最多讀取一次；額外空間取決於是否先儲存整條路徑，若直接累積答案字串則不需要再儲存座標清單。

### 題目規則（教學改寫）

字串長度是偶數（2～100），依序執行 `k` 個代碼：

| 代碼 | 操作 | 範例 |
|---|---|---|
| `0` | 每兩個字元交換 | `abcd → badc` |
| `1` | 每一對內部字典序排序 | `badc → abcd` |
| `2` | 把**前、後兩半**交錯組合 | `abcd → acbd` |

輸入是字串 `S`、整數 `k`、接著 `k` 個操作（官方範例每行一個）。輸出最後字串，只有一行。

````python
# === 練習題自動檢查工具：先執行，請不要修改 ===
def _check(title, func, cases):
    count = 0
    for i, (args, expected) in enumerate(cases, 1):
        try:
            got = func(*args)
            ok = got == expected
        except Exception as error:
            ok, got = False, f"發生錯誤：{type(error).__name__}: {error}"
        print(f"{'✅ PASS' if ok else '❌ 再試試'} | {title} 測試 {i}")
        if not ok:
            print(f"   輸入參數：{args!r}")
            print(f"   預期：{expected!r}")
            print(f"   實際：{got!r}")
        count += ok
    print(f"{title} 完成度：{count}/{len(cases)}\n")
    return count
````

**教師帶做**：`apcsntnu` 每兩字一組，將 `(ap)(cs)(nt)(nu)` 變成 `(pa)(sc)(tn)(un)`。

| Step | 配對前 | 配對後 |
|---|---|---|
| 1 | `(a,p)` | `(p,a)` |
| 2 | `(c,s)` | `(s,c)` |
| 3 | `(n,t)` | `(t,n)` |
| 4 | `(n,u)` | `(u,n)` |

**學生練習 2-1～2-5**
1. `abcd` 做一次 0。
2. `abcdefgh` 做一次 0。
3. `zz` 做一次 0。
4. 寫 `q2_swap_pairs(s)`，回傳轉換後的字串。
5. 對同一字串連做兩次 0，是否會還原？解釋原因。

````python
a, b = 1, 2
b, a = a, b

# 字串 -> list -> append -> join -> 字串
l = ['a', 'b', 'c']
print(' '.join(l))
````

````python
def q2_swap_pairs(s):
    # TODO：每對相鄰字元交換，回傳新的字串
    result = []
    for i in range(0, len(s), 2):
        result.append(s[i+1])
        result.append(s[i])
    return ''.join(result)
````

# Group 3 — 兩兩排序（操作 1）

**教師帶做**：`family` 分組 `(fa)(mi)(ly)`，每對分別排序得到 `(af)(im)(ly)`。

這是**每對各自排序**，不是先把整個字串排序。

**學生練習 3-1～3-5**
1. `dbca`：列出每組排序前後。
2. `zzzz`：結果是否變化？
3. `zbaa`：操作 1 的答案是什麼？
4. 寫 `q3_sort_pairs(s)`。
5. 操作 1 做兩次會和第一次結果相同嗎？為什麼？

````python
l = ['a', 'b']
l.extend(['c', 'd'])

print(l)
````

````python
def q3_sort_pairs(s):
    # TODO：每對字元分別排序，不能對整個字串做 sorted(s)
    result = []
    for i in range(0, len(s), 2):
        a, b = s[i], s[i+1]
        result.extend([a, b] if a <= b else [b, a])
    return ''.join(result)
````

````python
_check('Group 3', q3_sort_pairs, [(('family',), 'afimly'), (('zbaa',), 'bzaa'), (('dcbc',), 'cdbc')])
````

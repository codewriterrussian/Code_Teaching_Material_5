```python
def swap_pairs(s):
    result = []
    for i in range(0, len(s), 2):
        result.append(s[i+1])
        result.append(s[i])
    return ''.join(result)

def sort_pairs(s):
    result = []
    for i in range(0, len(s), 2):
        if s[i+1]>s[i]:
            result.append(s[i])
            result.append(s[i+1])
        else:
            result.append(s[i+1])
            result.append(s[i])
    return ''.join(result)

def interleave(s):
    result = []
    for i in range(len(s)//2):
        result.append(s[i])
        result.append(s[i+len(s)//2])
    return ''.join(result) 

for i in range(len(operations)):
    if operations[i] == 0:
        s = swap_pairs(s)
    elif operations[i] == 1:
        s = sort_pairs(s)
    elif operations[i] == 2:
        s = interleave(s)
```

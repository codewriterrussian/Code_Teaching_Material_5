判斷是否包含指定數字

寫一個函式：
def practice_contains(nums, target):
    return None

如果 target 出現在 nums 中，回傳 True，否則回傳 False。

Example:
practice_contains([2, 4, 6], 4)

應該回傳：
True

```python
def practice_contains(nums, target):

    flag = None

    for i in range(len(nums)):
        if nums[i]==target:
            return True
        else:
            flag = True

    if flag:
        return False

practice_contains([2, 4, 6, 4], 3)
```

```python
def q5_find_max(nums):
    max_value = 0

    # max_value 比較 list 的東西 -> 假如大於 max_value -> max_value = nums[idx]

    for i in range(len(nums)):
        if nums[i] > max_value:
            max_value = nums[i]

    return max_value

q5_find_max([3, 1, 8, 2])
```

```python
def q5_find_min(nums):

    min_value = nums[0]

    for i in range(1,len(nums)):
        if nums[i] < min_value:
            min_value = nums[i]

    return min_value

q5_find_min([5, 8, 2, 9])
```

lst = ['a', 2, 2.45, [1,2,3]]
def ...(lst) -> 分類 -> 不同類型就分成該類型的串列 (type) => [[str], [int], [float], [list]]

```python
def lst_type(x):

    str_l, int_l, float_l, list_l = [], [], [], []

    for i in range(len(x)):
        if type(x[i]) == str:
            str_l.append(x[i])
        elif type(x[i]) == int:
            int_l.append(x[i])
        elif type(x[i]) == float:
            float_l.append(x[i])
        elif type(x[i]) == list:
            list_l.append(x[i])
    
    return [str_l, int_l, float_l, list_l]

lst = ['a', 2, 2.45, [1,2,3],'cbd']
print(lst_type(lst))
```

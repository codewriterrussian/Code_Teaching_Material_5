# Lesson 6：輸出格式化與條件判斷

這份筆記整理成三個主題：
1. `print()` 的特殊字元：換行 `\n`、Tab `\t`
2. 數字與字串格式化：`%`、`.format()`、f-string
3. 條件判斷：比較運算、`if / elif / else`、巢狀判斷


## 1. `print()` 與特殊字元

`\n` 代表換行；`\t` 代表 Tab 空格，常用來做簡單表格對齊。


```python
print('輸出結果是： 你好\n')
print('hi	hi')
```

## 2. 舊式格式化：`%`

常見格式：
- `%s`：字串 string
- `%d`：整數 integer
- `%f`：浮點數 float
- `%.2f`：保留小數點後 2 位
- `%10.2f`：總寬度至少 10 格，小數點後 2 位


```python
name = 'Alice'
score = 95

print('Student Name: %s, Score: %d' % (name, score))
print('score: %d' % score)
```

### 小數位數控制

`1 / 3 = 0.33333333...`，可以用格式化控制要顯示幾位小數。


```python
output = 1 / 3

print('The value of 1/3 is approximately %10.10f' % output)
print('The value of 1/3 is approximately %10.2f' % output)
print('The value of 1/3 is approximately %5.2f' % output)
```

### 整數與浮點數格式差異

`%d` 會以整數方式輸出；`%f` 會以小數方式輸出。


```python
output = 9999 ** 2 + 0.005

print('The value of 9999^2 is approximately %d' % output)
print('The value of 9999^2 is approximately %2.5f' % output)
```

### 圓周長與面積範例

這裡練習把計算結果用固定小數位數印出。


```python
pi = 3.14159
radius = 5

circumference = 2 * pi * radius
area = pi * radius ** 2

print('The circumference of a circle with radius 5 is approximately %10.2f' % circumference)
print('The area of a circle with radius 5 is approximately %10.2f' % area)
```

```python
print('%.2f' % 3)
```

## 4. `.format()` 格式化

`.format()` 可以用 `{0}`、`{1}` 指定第幾個參數。

例：`'{0}-{1}-{0}'.format(A, B)` 會輸出 `A-B-A`。


```python
print('{0}-{1}-{0}'.format(A, B))
```

### 欄位寬度

`{0:3}` 表示第 0 個參數至少佔 3 格寬度。若內容不足，Python 會自動補空格。


```python
print('{0:3}{0:1}'.format(12, 345))   #  1212
print('{0:2}{1:3}'.format(12, 345))   # 12345
print('{0:3}{1:5}'.format(12, 345))   #  12  345
```

```python
print('{0:3}{0:1}'.format(12.3, 345.5))    # 12.312.3
print('{0:12}{1:3}'.format(12.3, 345.5))   #         12.3345.5
print('{0:3}{1:10}'.format(12.3, 345.5))   # 12.3     345.5
```

## 5. f-string 格式化

f-string 是現在 Python 很常用的格式化方式。  
`{score:10.2f}` 表示：總寬度至少 10 格，小數點後 2 位。


```python
score = 88.82

print(f'My score is {score:10.2f}')
```

請印出：名字、分數、電話。  
提示：可以使用 `\t` 讓欄位之間有 Tab 距離。

# 條件判斷

條件判斷可以讓程式根據「條件是否成立」決定要不要執行某段程式。


## 7. 比較運算

常見比較：
- `==`：是否相等
- `>`：是否大於
- `<`：是否小於
- `%`：取餘數，可用來判斷奇偶數


```python
x = 1

print(x == 1)
print(x > 10)
print(x % 2 == 0)
```

## 8. `if` 基本結構

如果條件成立，就執行縮排內的程式。


```python
x = 1

if x > 0:
    print('positive')

if x < 0:
    print('negative')
```

可以把 `if` 想成一道門：條件成立才能進去執行裡面的動作。

人 → 門（if 條件）→ 動作


```python
x = 0

if x == 0:
    print('zero')
    print('x is zero')
```

### 縮排很重要

只有縮排在 `if` 裡面的程式，才會受到條件控制。


```python
x = -1

if x > 0:
    print('A')

print('B')
```

### `True`、`False` 與 `not`

`not False` 會變成 `True`，所以第二個 `if` 也會執行。


```python
if True:
    print('This will always be printed.')

if not False:
    print('This will also be printed.')
```

```python
x = 1

if x == 1:
    print('x is 1')
```

## 9. `if / else`

`else` 代表：前面的條件不成立時，執行這裡。


```python
x = 2

if x > 0:
    print('positive')
else:
    print('non-positive')
```

## 10. `if / elif / else`

當情況有三種以上，可以使用 `elif`。


```python
x = int(input('Please enter a number: '))

if x > 0:
    print('positive')
elif x < 0:
    print('negative')
else:
    print('zero')
```

```python
score = 75

if score >= 90:
    print('A')
elif score >= 60:
    print('pass')
else:
    print('fail')
```

## 11. 巢狀條件判斷

巢狀條件判斷就是在 `if` 裡面再放一層 `if`，常用在更細的分類。


```python
score = int(input('Please enter your score: '))

if score > 100 or score < 0:
    print('Invalid score')
elif score >= 80:                           # A Range
    if score >= 90:
        print('A+')
    elif score >= 85:
        print('A')
    else:
        print('A-')
elif score >= 70:                           # B Range
    if score >= 77:
        print('B+')
    elif score >= 73:
        print('B')
    else:
        print('B-')
elif score >= 60:                           # C Range
    if score >= 67:
        print('C+')
    elif score >= 63:
        print('C')
    else:
        print('C-')
else:                                       # F Range
    print('fail')
```

### 練習 1：判斷奇偶數

如果 `x` 是偶數，印出 `x is even`；否則印出 `x is odd`。

## 1. 條件判斷的概念

可以把條件判斷想成「根據不同情況，做不同事情」。

例如：

```text
天氣
├─ 晴天 → 吃鼎泰豐
├─ 陰天 → 買麥當勞
└─ 雨天 → 叫外送
```

程式也可以使用 `if`、`elif`、`else` 表達這種決策流程。


## 2. 基本 `if`

如果條件成立，縮排在 `if` 底下的程式都會執行。

下面的例子中：

```python
x > 0
```

為 `True`，所以三個 `print()` 都會執行。


````python
x = 2

if x > 0:
    print("positive")
    print("positive")
    print("positive")
````

## 3. `if`、`elif`、`else`

當程式有多種可能情況時，可以使用：

```python
if 條件:
    ...
elif 其他條件:
    ...
else:
    ...
```

也可以在一個條件判斷裡再放另一個條件判斷，形成**巢狀條件判斷**。

### 成績等級判斷

下面的程式先判斷成績所在的大範圍，再使用第二層條件細分等級。

例如：

```text
score >= 80
    ↓
再判斷 90、85
    ↓
A+ / A / A-
```


````python
# if elif else
score = float(input())

if score >= 80:
    if score >= 90:
        print("A+")
    elif score >= 85:
        print("A")
    else:
        print("A-")
elif score >= 70:
    if score >= 77:
        print("B+")
    elif score >= 73:
        print("B")
    else:
        print("B-")
elif score >= 60:
    if score >= 67:
        print("C+")
    elif score >= 63:
        print("C")
    else:
        print("C-")
elif score > 0:
    print("F")
else:
    print("X")
````

## 6. 重點整理

### 基本條件

```python
if condition:
    ...
```

### 多種情況

```python
if condition1:
    ...
elif condition2:
    ...
else:
    ...
```

### 常見條件寫法

```python
x > 0
x % 2 == 0
len(password) >= 6
```

條件判斷最重要的是先確認：

1. 要判斷的條件是什麼？
2. 條件成立時要做什麼？
3. 條件不成立時要做什麼？

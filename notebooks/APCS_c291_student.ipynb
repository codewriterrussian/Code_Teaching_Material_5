**官方範例 2**

```text

```

輸出：`2`

````python
# === 練習題自動檢查工具：先執行，請不要修改 ===
def _check(title, func, cases):
    count = 0
    for i, (args, expected) in enumerate(cases, 1):
        try:
            got = func(*copy_args(args))
            ok = got == expected
        except Exception as error:
            ok, got = False, f"發生錯誤：{type(error).__name__}: {error}"
        print(f"{'✅ PASS' if ok else '❌ 再試試'} | {title} 測試 {i}")
        if not ok:
            print(f"   輸入參數：{args!r}")
            print(f"   預期：{expected!r}")
            print(f"   實際：{got!r}")
        count += ok
    print(f"{title} 完成度：{count}/{len(cases)}\n")
    return count

def copy_args(args):
    # 讓每個測試互不影響；Group 2、3 會修改 visited 清單。
    import copy
    return copy.deepcopy(args)
````

## 1. 程式的基本流程

程式通常可以分成三個部分：

```text
輸入 → 執行 → 輸出
```

例如，使用 ChatGPT 時：

```text
輸入指令 → AI 執行處理 → 顯示回應
```

演算法則是一組明確、有順序的解題步驟。

## 2. 使用 `print()` 顯示內容

`print()` 可以把文字、數字或其他資料顯示在畫面上。

````python
print("Hello, World!")
````

## 3. 單引號與雙引號

Python 可以使用單引號 `' '` 或雙引號 `" "` 表示字串。

兩種寫法都可以，但前後引號必須成對。

````python
print('j')
print("j")
````

## 4. 同時輸出多個值

在 `print()` 中使用逗號分隔多個值時，Python 預設會在它們之間加入一個空格。

````python
print(123, "早安", "a", 456)
````

## 5. 使用 `sep` 設定分隔符號

`sep` 是 separator（分隔符號）的縮寫。

它可以控制多個輸出值之間要放入什麼內容。

````python
print(1, 2, 3, 4, 5)
print(1, 2, 3, 4, 5, sep=" ")
print(1, 2, 3, 4, 5, sep=",")
````

## 6. 使用 `end` 設定輸出結尾

`print()` 預設會在輸出後換行。

`end` 可以改變輸出結尾的內容。預設值是換行符號 `\n`。

````python
print(123, end=" ")
print(456)
````

## 7. 使用 `input()` 接收輸入

`input()` 會暫停程式，等待使用者輸入資料。

括號中的文字是提示訊息。使用者輸入的資料會以字串形式傳回。

````python
user_text = input("請輸入內容：")
print("你輸入的是：", user_text)
````

## 8. 接收多個輸入值

為了讓程式更容易閱讀，可以先把每次輸入存入變數，再使用 `print()` 顯示。

````python
first_value = input("請輸入第一個值：")
second_value = input("請輸入第二個值：")

print(first_value, second_value)
````

## 9. 綜合範例：姓名與年齡

以下程式先接收姓名與年齡，再將完整句子顯示在同一行。

````python
name = input("請輸入你的名字：")
age = input("請輸入你的年齡：")

print("你好", name, "，你今年", age, "歲。", sep="")
````

## 10. 使用 `split()` 分割同一行輸入

`split()` 可以按照空白將一個字串分割成多個部分。

例如，輸入：

```text
小明 13 台北
```

可以分別存入 `name`、`age` 和 `city`。

````python
name, age, city = input("請依序輸入姓名、年齡與城市：").split()

print("你好", name, "，你今年", age, "歲，你住在", city, "。", sep="")
````

## 重點整理

- `print()`：顯示資料。
- `sep`：設定多個輸出值之間的分隔符號。
- `end`：設定輸出的結尾。
- `input()`：接收使用者輸入。
- `split()`：將一行文字分割成多個部分。
- 使用變數保存輸入，可以讓程式更清楚、更容易閱讀。

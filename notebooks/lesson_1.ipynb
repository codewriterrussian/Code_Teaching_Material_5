程式： 輸入 -> 執行 -> 輸出
=> ChatGPT: 輸入指令 -> AI 運行 -> 給你回應
=> ChatGPT -> 裡面的小東西/小程式 => 輸入 -> 執行 -> 輸出 (演算法：明確的架構)

```python
print("Hello, World!")  # 輸入->"Hello, World!" / 執行->print
# 輸出在下面👇
```

```python
print('j') # ' 跟 " 都可以使用
```

```python
print(123, '早安', 'a', 456) # 可以同時輸出多個值，並且會自動在中間加上空格
```

```python
print(1, 2, 3, 4, 5)
print(1, 2, 3, 4, 5, sep=' ')
print(1, 2, 3, 4, 5, sep=',')
```

```python
# 我們在寫print的時候，通常會需要在同一行輸出值 -> (輸出 + 換行) 這樣的情況。
# 這時候就可以使用 end 參數來控制輸出的結尾符號，預設是換行符號 \n。
print(123, end=' ')
print(456)
```

```python
# input -> 把人類輸入的東西藏在電腦裡面，假如我們沒有做輸出的動作的話，他就不會有結果
input('提示： ') # 括號裡面的字就是提示訊息，會在輸入前顯示給使用者看
print(input()) 
```

```python
# 測試：輸出 1,2,3,4,5
print(input(), input())
```

```python
# 加入顯示年齡 -> 都寫在一行
print('你好', input('請輸入你的名字：'), '你今年', input('請輸入你的年齡：'), '歲')

##

print('你好', input('請輸入你的名字：'), end='，') # end 參數可以控制輸出的結尾符號，預設是換行符號 \n。
print('你今年', input('請輸入你的年齡：'), '歲')
```

```python
name, age, city = input().split()
# 小明 13 台北 -> ['小明', '13', '台北']  
# # split() -> 把字串切割成一個串列 (資料型態)

print('你好', name, '你今年', age, '歲', '你住在', city)
```

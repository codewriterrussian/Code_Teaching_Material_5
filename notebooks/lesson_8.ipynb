```python
lst = [3, 5, 2, -1, 10]
count = 0

for i in lst:
    if i < 0:
        break
    count += i

print(count)
```

```python
lst = [3, 5, 2, -1, 10]
count = 0
idx = 0

while lst[idx] >= 0:
    count += lst[idx]
    idx += 1

print(count)
```

回傳平方串列

寫一個函式：

def practice_square_list(nums):
    return None

回傳一個新的串列，裡面每個數字都是原本數字的平方。

Example:
practice_square_list([1, 2, 3])
practice_square_list([5, 9, 4])

應該回傳：
[1, 4, 9]

```python
def practice_square_list(nums):

    for i in range(len(nums)):
        nums[i]= nums[i]**2

    return nums
```

```python
practice_square_list([5, 9, 4, 1, 2, 3])
```

找出使用者輸入數值的因數

完成函式：
def q9_find_factors(n):
    #TODO: 回傳 n 的所有正因數，並存成串列
    return None

Example:
q9_find_factors(12)

應該回傳：
[1, 2, 3, 4, 6, 12]

提示：如果 n % i == 0，代表 i 是 n 的因數。

```python
# 12 % 1
# 12 % 2
# ... 12 %12

def q9_find_factors(n):
    lst = []

    for i in range(1,n+1):
        if n % i == 0:
            lst.append(i)

    return lst

q9_find_factors(12)
```

計算字串中有幾個指定字母

完成函式：
def q10_count_char(text, target):
    #TODO: 回傳 target 在 text 中出現幾次
    return None

Example:
q10_count_char("banana", "a")

應該回傳：
3

```python
def q10_count_char(text, target):
    count = 0

    for i in text:
        if i == target:
            count += 1

    return count

q10_count_char("banana", "a")
```

## 2. 把加法寫成函式

函式可以把重複使用的程式碼包起來。  
下面的 `add_sum()` 會接收兩個數字，轉成整數後回傳相加結果。

````python
def add_sum(num1, num2):
    num1, num2 = int(num1), int(num2)
    return num1 + num2
````

## 3. 函式的基本概念

可以把函式想成一個小機器：

```text
輸入 input  →  函式內部處理 process  →  輸出 output
```

例如：  
`double(6)` 的輸入是 `6`，函式內部做 `6 * 2`，最後回傳 `12`。

````python
def double(x):
    return x * 2

print(double(6))
````

````python
b = 'apple'

def out(a):
    print(a)

out(b)
````

## 6. 商數與餘數：`//` 和 `%`

Python 中：

- `//` 表示整數除法，得到「商數」
- `%` 表示取餘數

例如：

```python
5 // 2  # 2
5 % 2   # 1
```

下面的函式會先確認哪個數比較大，再用大的數除以小的數。

````python
def two_num_divide(num1, num2):
    num1, num2 = int(num1), int(num2)

    # 避免除以 0
    if num1 == 0 or num2 == 0:
        return "不能除以 0"

    # 確保 num1 是較大的數
    if num1 < num2:
        num1, num2 = num2, num1

    # 計算商數與餘數
    quotient = num1 // num2
    remainder = num1 % num2

    return quotient, remainder
````

````python
quotient, remainder = two_num_divide(5, 2)

print("商數:", quotient)
print("餘數:", remainder)
````

````python
# 你也可以試試看兩個數字順序相反的情況
two_num_divide(2, 5)
````

## 7. 指數函式：`base ** exp`

如果要計算：

```text
2^3 = 8
```

其中 `2` 是底數 base，`3` 是指數 exponent。

下面的函式會接收底數和指數，並回傳指數運算結果。

````python
def exponent(base, exp):
    base, exp = int(base), int(exp)
    return base ** exp
````

````python
exponent(5, 6)
````

## 9. 自己寫出 `sum`、`min`、`max`

下面的函式會自己用迴圈完成三件事：

1. 加總全部數字  
2. 找出最小值  
3. 找出最大值  

這是一個很重要的練習，因為它讓你理解內建函式背後的邏輯。

````python
def list_function(l):
    # 檢查輸入是否為 list
    if type(l) != list:
        return "The input is not a list"

    # 檢查 list 是否為空
    if l == []:
        return "The input is an empty list"

    num_sum = 0
    num_min = None
    num_max = None

    for ele in l:
        # 更新最小值
        if num_min is None or ele < num_min:
            num_min = ele

        # 更新最大值
        if num_max is None or ele > num_max:
            num_max = ele

        # 更新總和
        num_sum += ele

    return num_sum, num_min, num_max
````

````python
list_function([3, 8, 2, 10, 100])
````

## 10. 判斷直角三角形

給三個邊長，我們想判斷它是不是直角三角形。  

根據畢氏定理：

```text
最長邊² = 另外兩邊² 的和
```

也就是：

```text
c² = a² + b²
```

如果符合，就可以計算：

```text
sin = 對邊 / 斜邊
cos = 鄰邊 / 斜邊
```

在這個簡化版本中，我們把排序後最短的邊當作 `sin` 的分子，第二長的邊當作 `cos` 的分子。

````python
def check_tri(lengths):
    # 確認有三個邊長
    if len(lengths) != 3:
        return "這不是我要的東西：請輸入三個邊長"

    # 由小到大排序
    lengths = sorted(lengths)
    length_1, length_2, length_3 = lengths

    # 判斷是否符合畢氏定理
    if length_3 ** 2 == length_2 ** 2 + length_1 ** 2:
        sin_out = length_1 / length_3
        cos_out = length_2 / length_3
        return sin_out, cos_out
    else:
        return "這不是我要的東西：不是直角三角形"
````

````python
check_tri([3, 4, 5])
````

````python
check_tri([2, 3, 4])
````
